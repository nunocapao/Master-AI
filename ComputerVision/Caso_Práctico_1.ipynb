```
import cv2
from google.colab.patches import cv2_imshow

# Ruta a la imagen
ruta_imagen = '/content/drive/MyDrive/IA_Mestrado/imagen_formas_simples.png'

# Leer la imagen
imagen = cv2.imread(ruta_imagen)

# Verificar si se cargó correctamente
if imagen is None:
    print('Error: No se pudo cargar la imagen. Verifica el nombre o la ruta del archivo.')
else:
    # Mostrar la imagen
    cv2_imshow(imagen)
```

```
# Convertir la imagen a escala de grises
imagen_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

# Mostrar la imagen en escala de grises
cv2_imshow(imagen_gris)
```

```
# Usamos GaussianBlur con un tamaño de kernel de 5x5
imagen_desenfocada = cv2.GaussianBlur(imagen_gris, (5, 5), 0)

# Mostrar la imagen desenfocada
cv2_imshow(imagen_desenfocada)
```

```
# Usamos Canny con dos umbrales: mínimo y máximo
bordes = cv2.Canny(imagen_desenfocada, 50, 150)

# Mostrar la imagen con los bordes detectados
cv2_imshow(bordes)
```

```
# Guardar imagens
cv2.imwrite('/content/drive/MyDrive/IA_Mestrado/imagen_gris.png', imagen_gris)
cv2.imwrite('/content/drive/MyDrive/IA_Mestrado/imagen_desenfocada.png', imagen_desenfocada)
cv2.imwrite('/content/drive/MyDrive/IA_Mestrado/imagen_bordes.png', bordes)

print("Imágenes guardadas correctamente.")
```
